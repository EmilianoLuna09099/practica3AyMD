En este archivo se hace la clasificacion de variables para entender si son cuantitativas o cualitativas.

Observando con detenimiento los datos, la mayoria de los datos son cualitativos. Sin embargo viendo con detenimiento los pocos que son cuantitativos parecen ser discretos, asi que solo para corroborar esto, esta la celda de abajo. 

```python
import polars as pl
import sys

sys.path.append("..")

from config.rutas import RUTA_DATOS_CRUDOS

df = pl.read_csv(RUTA_DATOS_CRUDOS)

df.filter(
    pl.col("ingreso_pareja").mod(1) != 0
).select("ingreso_pareja")

df.filter(
    pl.col("factor_expansion").mod(1) != 0
).select("factor_expansion")
```

| Variable                    | Cualitativa / Cuantitativa | Tipo     | Justificacion | 
| --------------------------- | -------------------------- | -------- | ----- | 
| `cve_entidad`               | Cualitativa                | Nominal  | Es una clave identificadora; el número no representa que la entidad tenga algun valor.    |
| `nom_entidad`               | Cualitativa                | Nominal  | Es el nombre de la entidad y no tiene un orden.                       |
| `cve_municipio`             | Cualitativa                | Nominal  | Es una clave identificadora del municipio, de igual manera no representa que la entidad tenga algun valor.                            |
| `nom_municipio`             | Cualitativa                | Nominal  | Es el nombre del municipio y no tiene un orden.                       |
| `edad_primer_union`         | Cualitativa               | Nominal | Representa la respuesta de la opciones diponibles, por ejemplo, 0 es no, 1 es si, 98 es no se. Por esta razon esta columna tiene muchos espacios en blanco tambien.                   |
| `num_hijos`                 | Cuantitativa               | Discreta | Es un conteo y solamente puede tomar valores enteros.                 |
| `nivel_escolaridad`         | Cualitativa                | Ordinal  | Sus categorías representan niveles educativos con un orden.           |
| `estado_civil_id`           | Cualitativa                | Nominal  | Es un código numérico que representa una de las opciones disponibles.                   |
| `estado_civil_desc`         | Cualitativa                | Nominal  | Representa categorías de estado civil pero al contrario que la anterior, esta describe con una cadena el valor.                      |
| `estrato_socioeconomico`    | Cualitativa                | Ordinal  | Representa categorías socioeconómicas que pueden ordenarse.           |
| `pareja_trabaja_id`         | Cualitativa                | Nominal  | Es un código que representa una de las dos posibles categorías, si trabaja o no.                            |
| `pareja_trabaja_desc`       | Cualitativa                | Nominal  | Es igual el codigo que representa una de las dos posibles categorias, pero esta vez en una cadena.                           |
| `ingreso_pareja`            | Cuantitativa               | Discreta | Representa una cantidad monetaria en la cual todas las cantidades estan en numero cerrados.          |
| `dinero_propio_id`          | Cualitativa                | Nominal  | Es un código que identifica una categoría.                            |
| `dinero_propio_desc`        | Cualitativa                | Nominal  | Describe categorías sobre la disponibilidad de dinero propio en una cadena.         |
| `apoyo_gobierno_id`         | Cualitativa                | Nominal  | Es un código que identifica una categoría, si se le otorga apoyo del gobierno o no.                            |
| `apoyo_gobierno_desc`       | Cualitativa                | Nominal  | Describe si se le otorga apoyo del gobierno o no, en forma de cadena.          |
| `tiene_ahorros_id`          | Cualitativa                | Nominal  | Es un código que identifica una categoría, si tiene ahorros o no.                            |
| `tiene_ahorros_desc`        | Cualitativa                | Nominal  | Describe si la persona cuenta o no con ahorros, en una cadena. |
| `propietaria_vivienda_id`   | Cualitativa                | Nominal  | Es un código que identifica una categoría, si es propietaria de la vivienda o no.                            |
| `propietaria_vivienda_desc` | Cualitativa                | Nominal  | Describe la situación de propiedad de la vivienda, en cadena.                    |
| `sufrio_violencia_pareja`   | Cualitativa                | Nominal  | Representa  con codigo si sufrio algun tipo de violencia o no.              |
| `factor_expansion`          | Cuantitativa               | Discreta | Indica a cuántas unidades reales de la población total representa esa observación.          |
| `anio_encuesta`             | Cuantitativa               | Discreta | Representa el año mediante un valor entero.                           |
